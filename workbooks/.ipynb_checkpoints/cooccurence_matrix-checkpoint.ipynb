```python
import pandas as pd
import itertools
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
pcanc_tcga = pd.read_csv('/SAN/colcc/Hormad1/data/clinical/pcanc_cohort.csv')
pcanc_tcga.head()
```

```python
pcanc_tcga.columns
```

```python
SURVIVAL_POINT = "60_months"
```

```python
pcanc_tcga_60 = pcanc_tcga[pcanc_tcga[SURVIVAL_POINT]!='Censored']
pcanc_tcga_60 = pcanc_tcga_60[pcanc_tcga_60[SURVIVAL_POINT]!='Unknown']
pcanc_tcga_60 = pcanc_tcga_60[~pcanc_tcga_60[SURVIVAL_POINT].isna()]
```

```python
pcanc_tcga_60['stage_var'] = 'Stage' + pcanc_tcga_60['stage_var'].astype(str)
pcanc_tcga_60['age_group'] = 'Age' + pcanc_tcga_60['age_group'].astype(str)
pcanc_tcga_60['project_id_x'] = pcanc_tcga_60['project_id_x'].astype(str)
pcanc_tcga_60[SURVIVAL_POINT] = pcanc_tcga_60[SURVIVAL_POINT].astype(str)  # e.g., 'survived' or 'died'
```

```python
df_expanded = pcanc_tcga_60.apply(lambda row: [row['stage_var'], row['age_group'], row['project_id_x'], row[SURVIVAL_POINT]], axis=1)

pair_counts = {}
for row in df_expanded:
    for a, b in itertools.combinations(sorted(row), 2):
        key = tuple(sorted([a, b]))
        pair_counts[key] = pair_counts.get(key, 0) + 1

categories = sorted(set(itertools.chain.from_iterable(df_expanded)))
co_matrix = pd.DataFrame(0, index=categories, columns=categories)

for (a, b), count in pair_counts.items():
    co_matrix.loc[a, b] = count
    co_matrix.loc[b, a] = count

for cat in categories:
    co_matrix.loc[cat, cat] = df_expanded.apply(lambda row: cat in row).sum()

plt.figure(figsize=(12, 10))
sns.heatmap(co_matrix, annot=True, fmt='d', cmap='Blues')
plt.title("Full Co-occurrence Matrix (Stage, Age, Cancer, Survival)")
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
pcanc_tcga['stage_var'] = 'Stage' + pcanc_tcga['stage_var'].astype(str)
```

```python
df_survived = pcanc_tcga[pcanc_tcga['60_months'] == 'Alive']
df_died = pcanc_tcga[pcanc_tcga['60_months'] == 'Died']

def build_co_matrix(sub_df):
    data_rows = sub_df.apply(lambda row: [row['stage_var'], row['age_group'], row['project_id_x']], axis=1)
    pair_counts = {}
    for row in data_rows:
        for a, b in itertools.combinations(sorted(row), 2):
            key = tuple(sorted([a, b]))
            pair_counts[key] = pair_counts.get(key, 0) + 1

    # Unique labels
    categories = sorted(set(itertools.chain.from_iterable(data_rows)))

    # Build square matrix
    matrix = pd.DataFrame(0, index=categories, columns=categories)
    for (a, b), count in pair_counts.items():
        matrix.loc[a, b] = count
        matrix.loc[b, a] = count

    # Diagonal (self co-occurrence)
    for cat in categories:
        matrix.loc[cat, cat] = data_rows.apply(lambda row: cat in row).sum()

    return matrix

# Build the two matrices
co_matrix_survived = build_co_matrix(df_survived)
co_matrix_died = build_co_matrix(df_died)

# Plot side by side
plt.figure(figsize=(30, 12))

plt.subplot(1, 2, 1)
sns.heatmap(co_matrix_survived, annot=True, fmt='d', cmap='Blues')
plt.title('Co-occurrence (Survived)')
plt.xticks(rotation=90)
plt.yticks(rotation=0)

plt.subplot(1, 2, 2)
sns.heatmap(co_matrix_died, annot=True, fmt='d', cmap='Reds')
plt.title('Co-occurrence (Died)')
plt.xticks(rotation=90)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```
