```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1) helper: counts per patient in a slice
def counts_per_patient(df: pd.DataFrame) -> pd.DataFrame:
    counts = (
        df.groupby(["patient_id", "tissue_class", "attention_group"])
          .size()
          .rename("count")
          .reset_index()
    )

    grp = ['patient_id', 'attention_group']
    
    # add percentage within each patient × attention_group
    counts['pct'] = (
        counts['count'] /
        counts.groupby(grp)['count'].transform('sum')
        * 100
    ).round(2)
    
    # tidy result (one row per patient_id × attention_group × tissue_class)
    pct_long = counts[['patient_id','attention_group','tissue_class','count','pct']]
    return pct_long


# 2) single-cancer plotting function: one row with up to 4 concept subplots, log y-axis
def plot_concepts_for_cancer(
    df: pd.DataFrame,
    cancer_type: str,
    surv_keep=("1_TP", "0_TN"),
    concepts=("Cancer", "Stage", "RNA_Bio_bin", "Age"),
    log_y: bool = True,
    show: bool = True,
):
    """
    For a given cancer_type, create a single-row figure with up to four subplots (one per concept).
    Each subplot: boxplot of per-patient counts by tissue_class (x) and attention_group (hue).
    Optionally log-scale the y-axis to avoid one concept blowing the linear scale.
    """
    # filter to this cancer and selected survival results
    sub = df[(df["cancer_type"] == cancer_type) & (df["surv_result"].isin(surv_keep))].copy()
    if sub.empty:
        raise ValueError(f"No rows after filtering for cancer_type={cancer_type} and surv_result in {surv_keep}.")

    # concepts to plot: keep requested order if present; otherwise take up to 4 present concepts
    present_concepts = list(pd.Series(sub["concept"].dropna().unique()).sort_values())
    requested = [c for c in concepts if c in present_concepts]
    concepts = (requested if requested else present_concepts)[:4]
    if len(concepts) == 0:
        raise ValueError(f"No matching concepts found in data for cancer_type={cancer_type}.")

    # consistent orders across subplots
    class_order = list(pd.Series(sub["tissue_class"].dropna().unique()).sort_values())
    att_groups = list(pd.Series(sub["attention_group"].dropna().unique()).sort_values())
    preferred = ["Low", "Medium", "High"]
    hue_order = preferred if set(preferred).issubset(att_groups) else att_groups

    # layout: single row, exactly 4 axes (hide extras if <4 concepts present)
    n = len(concepts)
    ncols = 4
    fig_width = max(14, 1.1 * len(class_order) * ncols)
    fig, axes = plt.subplots(1, ncols, figsize=(fig_width, 5), sharey=False)
    axes = list(np.atleast_1d(axes))

    # hide unused axes if fewer than 4 concepts present
    for k in range(n, ncols):
        axes[k].set_visible(False)

    # track y-range for harmonization (esp. when log)
    y_max = 0
    y_min_pos = np.inf

    for ax, concept in zip(axes, concepts):
        slice_c = sub[sub["concept"] == concept]
        counts = counts_per_patient(slice_c)
        if counts.empty:
            ax.set_visible(False)
            continue

        sns.boxplot(
            data=counts,
            x="tissue_class",
            y="pct",
            hue="attention_group",
            order=class_order,
            hue_order=hue_order,
            showfliers=False,
            ax=ax,
        )

        ax.set_title(str(concept))
        ax.set_xlabel("tissue_class")
        ax.set_ylabel("% tissue_class")
        ax.tick_params(axis="x", rotation=30)

        # optional log scaling to handle large ranges
        #if log_y:
            #ax.set_yscale("log")

        # update y-range trackers
        if not counts["pct"].empty:
            y_max = max(y_max, counts["pct"].max())
            min_pos = counts["pct"][counts["pct"] > 0].min()
            if pd.notna(min_pos):
                y_min_pos = min(y_min_pos, min_pos)

    # tidy legend: keep one global legend
    handles, labels = None, None
    for ax in axes[:n]:
        leg = ax.get_legend()
        if leg is not None:
            handles, labels = ax.get_legend_handles_labels()
            leg.remove()
    if handles and labels:
        fig.legend(handles, labels, title="attention_group", loc="upper right", bbox_to_anchor=(0.995, 0.995))

    # harmonize y-limits across visible axes
    for ax in axes[:n]:
        if not ax.get_visible():
            continue
        if log_y:
            low = max(1, int(y_min_pos)) if np.isfinite(y_min_pos) else 1
            high = y_max * 1.2 if y_max > 0 else None
            if high:
                ax.set_ylim(low, high)
        else:
            ax.set_ylim(0, y_max * 1.1 if y_max > 0 else 1)

    fig.suptitle(f"{cancer_type}: % tissue class per patient and attention group", y=0.99, fontsize=14)
    fig.tight_layout(rect=[0, 0, 1, 0.95])

    if show:
        plt.show()
    return fig


```

```python
def plot_all_cancers(
    df: pd.DataFrame,
    surv_keep=("1_TP", "0_TN"),
    concepts=("Cancer", "Stage", "RNA_Bio_bin", "Age"),
    log_y: bool = True,
    show: bool = True,
):
    """
    Loops over each cancer_type and produces one figure per cancer type, with up to
    four concept subplots in a single row. Returns a dict of {cancer_type: Figure}.
    """
    figs = {}
    for ct in sorted(pd.Series(df["cancer_type"].dropna().unique())):
        try:
            figs[ct] = plot_concepts_for_cancer(
                df,
                ct,
                surv_keep=surv_keep,
                concepts=concepts,
                log_y=log_y,
                show=show,
            )
        except ValueError as e:
            # skip cancer types with no usable data under the filters
            print(f"[skip] {ct}: {e}")
            continue
    return figs
```
